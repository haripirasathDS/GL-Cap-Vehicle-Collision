```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.linear_model import LogisticRegression

# 1. DATA ACCESS [cite: 23, 69]

github_url = https://raw.githubusercontent.com/haripirasathDS/GL-Cap-Vehicle-Collision/main/Dataset/Motor_Vehicle_Collisions_-_Crashes
dfa = pd.read_csv(github_url)

# 2. DATA CLEANING LOG
# Handling missing values in key predictors
dfa['BOROUGH'] = dfa['BOROUGH'].fillna('UNKNOWN')
dfa['CONTRIBUTING_FACTOR_1'] = dfa['CONTRIBUTING_FACTOR_1'].fillna('Unspecified')

# 3. FEATURE ENGINEERING: TIME CATEGORIES (RQ4)
dfa['CRASH_TIME'] = pd.to_datetime(dfa['CRASH_TIME'])
dfa['TIME_WINDOW'] = pd.cut(dfa['CRASH_TIME'].dt.hour,
                            bins=[0, 6, 12, 18, 24],
                            labels=['Night', 'Morning', 'Afternoon', 'Evening'],
                            right=False)

# 4. EDA: BOROUGH-BASED RISK (RQ2)
plt.figure(figsize=(12, 6))
sns.countplot(data=dfa, x='BOROUGH', hue='TIME_WINDOW')
plt.title('Collision Frequency by Borough and Time of Day')
plt.savefig('borough_eda.png') # For report inclusion
plt.show()

# 5. MODELLING PLACEHOLDER (RQ1 & RQ3)
print("EDA Complete. Proceeding to Logistic Regression and K-Means in the next phase.")
```
